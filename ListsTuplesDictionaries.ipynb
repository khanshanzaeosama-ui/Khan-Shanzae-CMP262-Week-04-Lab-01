## Part 3 – Dictionaries

A **dictionary** stores data using **key-value pairs**.

### Reminder
Use the pattern shown earlier in class. Do not copy an answer from the README.

```python

    "name": "Alex",
    "age": 20
}

print(person)

```

A key works like a label. Here, `"name"` is the key and `"Alex"` is the value.

````python
person = {
    "name": "Alex",
    "age": 20
}

print(person)
print(person["name"])
````
